```python
# Install the Google GenAI SDK
!pip install -q -U google-genai

import os
from google.colab import userdata
from google.genai import types
from google.genai import client

# Retrieve Gemini API Key from Google Colab Secrets
# Make sure you have added your API key under Colab's Key/Secrets tab as 'GEMINI_API_KEY'
api_key = userdata.get('GEMINI_API_KEY')

# Initialize the Gemini Client
ai = client.Client(api_key=api_key)

print("Setup complete! Gemini client is initialized.")
```

```python
def generate_text(prompt, system_instruction=None, temperature=1.0, top_p=0.95, max_output_tokens=1000):
    """
    Helper function to generate text using Gemini 2.5 Flash.
    Includes max_output_tokens to keep responses concise.
    """
    config = types.GenerateContentConfig(
        system_instruction=system_instruction,
        temperature=temperature,
        top_p=top_p,
        max_output_tokens=max_output_tokens
    )

    response = ai.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt,
        config=config
    )

    return response.text

print("Helper function updated with max_output_tokens.")
```

```python
# Topic for generation
topic = "An ancient clocktower in the heart of a bustling city that suddenly stopped ticking."

# Define system instructions for three distinct personas
roles = {
    "Poet": "You are an expressive poet. Write in vivid imagery, metaphors, and evocative poetic style.",
    "Journalist": "You are a concise news journalist. Write in an objective, factual, breaking-news news report style.",
    "Teacher": "You are an engaging elementary school teacher. Explain the event simply, warmly, and with educational curiosity."
}

prompt = f"Write about this event: {topic}"

# Generate outputs for each role
for role_name, persona_instruction in roles.items():
    print(f"=== ROLE: {role_name.upper()} ===")
    response = generate_text(
        prompt=prompt,
        system_instruction=persona_instruction,
        temperature=0.7
    )
    print(response)
    print("\n" + "="*50 + "\n")
```
